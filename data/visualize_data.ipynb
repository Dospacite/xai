```python
import json

def load_json_file(file_path):
    """Load and return the contents of a JSON file."""
    with open(file_path, 'r') as file:
        data = json.load(file)
    return data

def save_json_file(data, file_path):
    """Save data to a JSON file."""
    with open(file_path, 'w') as file:
        json.dump(data, file)
```

```python
import requests
import gzip

url = "http://data.phishtank.com/data/online-valid.json.gz"
response = requests.get(url)
with open("verified_online.json.gz", "wb") as f:
    f.write(response.content)


with gzip.open("verified_online.json.gz", "rb") as f_in:
    with open("verified_online.json", "wb") as f_out:
        f_out.writelines(f_in)

# remove the compressed file after extraction
import os
os.remove("verified_online.json.gz")
```
